```python
import pandas as pd
import random

random.seed(42)

# 1. UTILITIES
utility_names = [
    "ECG", "NEDCo", "VRA", "GRIDCo", "Enclave Power",
    "Meridian", "Sunon Asogli", "Bui Power Authority",
    "Karpowership Ghana", "AMERI Energy"
]
regions = ["Greater Accra", "Ashanti", "Eastern", "Western",
           "Central", "Volta", "Northern", "Bono", "Ahafo", "Oti"]

utilities = pd.DataFrame({
    "utility_id": [f"U{i+1:02d}" for i in range(10)],
    "utility_name": utility_names,
    "region": regions,
    "type": [random.choice(["Generation", "Distribution", "Transmission"]) for _ in range(10)]
})

# 2. SUBSTATIONS
substations = pd.DataFrame({
    "substation_id": [f"S{i+1:02d}" for i in range(44)],
    "substation_name": [f"Substation {i+1}" for i in range(44)],
    "utility_id": [random.choice(utilities["utility_id"]) for _ in range(44)],
    "region": [random.choice(regions) for _ in range(44)],
    "capacity_mva": [random.choice([25, 50, 75, 100, 150, 200]) for _ in range(44)],
    "latitude": [round(random.uniform(4.5, 11.0), 4) for _ in range(44)],
    "longitude": [round(random.uniform(-3.3, 1.2), 4) for _ in range(44)],
})

# 3. LINES (connections between substations)
lines = []
sub_ids = substations["substation_id"].tolist()
for i in range(55):
    a, b = random.sample(sub_ids, 2)
    lines.append({
        "line_id": f"L{i+1:03d}",
        "from_substation": a,
        "to_substation": b,
        "voltage_kv": random.choice([11, 33, 69, 161, 330]),
        "length_km": round(random.uniform(2, 80), 1),
        "status": random.choice(["Active", "Active", "Active", "Under Maintenance"])
    })
lines = pd.DataFrame(lines)

# SAVE TO CSV
utilities.to_csv("utilities.csv", index=False)
substations.to_csv("substations.csv", index=False)
lines.to_csv("lines.csv", index=False)

print("Files created successfully!")
print(utilities.shape, substations.shape, lines.shape)
```
